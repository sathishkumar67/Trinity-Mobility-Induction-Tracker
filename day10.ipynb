```python
def get_sentiment(text, examples=[]):

    prompt = "Classify sentiment:\n"
    for ex in examples:
        prompt += f"Text: {ex[0]}\nSentiment: {ex[1]}\n"
    prompt += f"Text: {text}\nSentiment:"
    return "Positive" 

print(get_sentiment("I love this!"))
print(get_sentiment("This is okay.", [("Bad movie", "Negative"), ("Great food", "Positive")]))
```
