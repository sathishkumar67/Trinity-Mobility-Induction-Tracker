# Chain of Thought
## Prompt: 
A store has 50 apples. A customer buys 15 apples. The store then receives a new shipment of 40 apples. The store manager takes 20% of the total current apples home. How many apples are left in the store? Let's think step by step.

## CoT Output:
1. The store starts with 50 apples.
2. Customer buys 15 apples. 50 - 15 = 35 apples
3. The store receives a shipment of 40 apples. $35 + 40 = 75$ apples total.
4. The manager takes 20% of the 75 apples. $75 * 0.20 = 15$ apples taken.
5. Subtract the manager's apples from the total. $75 - 15 = 60$ apples left.
6. There are 60 apples left in the store.

# Self-Consistency and Tree of Thoughts
## Prompt:
Evaluate if this startup idea is viable: "An app that translates dog barks into English." Generate three completely different analytical perspectives (Technical, Market, and Psychological). Then, based on a majority vote of those three perspectives, give a final "Viable" or "Not Viable" verdict.

## Output:
* path1: Current AI systems can detect emotional states but cannot map barks to complex human language because dogs do not possess a structured linguistic grammar. Verdict: Not Viable.
* path2: Pet owners spend more money on pet tech. The novelty factor alone would drive massive downloads and subscription revenue. Verdict: Viable.
* path3: This can lead to dangerous misunderstandings of a dog's actual needs. Verdict: Not Viable.

Majority Vote: Not Viable
