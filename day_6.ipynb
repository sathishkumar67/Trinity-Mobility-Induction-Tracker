```python
import torch
from transformers import pipeline

model_id = "meta-llama/Llama-3.2-1B"

pipe = pipeline(
    "text-generation", 
    model=model_id, 
    torch_dtype=torch.bfloat16, 
    device_map="auto",
)

stop_words = ["\n\n", "User:", "THE END"]

output = pipe(
    "The key to life is",
    max_new_tokens=100,       
    temperature=0.7,          
    top_p=0.9,               
    top_k=50,                 
    do_sample=True,           
    stop_strings=stop_words,  
    tokenizer=pipe.tokenizer,
    truncation=True           
)

print(output[0]['generated_text'])
```
